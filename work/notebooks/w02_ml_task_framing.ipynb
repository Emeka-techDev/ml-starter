## 3. Success metric


*One metric you can defend. What number means 'good'?*

The primary success metric will be **Precision@k** particularly  Precision@20
> This describes the number of pages that are actually declining according to the proxy target.
 For example a precision@20 of 0.3 means that 6 out of the 20 ranked pages are classified as declining.

 The Average precision will also be used as a secondary metric to evaluate the quality of the ranking beyond just the top 20

 A good result is therefore not a high model accuracy. It is a model that places substantially higher proportion of relevant review candidates in the top of the queue than the baseline, while remaining useful and explanable enough to support a content decision
